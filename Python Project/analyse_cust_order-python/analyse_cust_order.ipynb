```python
# -----------------------------------------------------------
print("\033[1mProject: Analyzing Customer Orders Using Python\033[0m")
print("-" * 50)
# -----------------------------------------------------------


#Task:1
#Store customer orders
# Create a list of customer names
customer_name =["Julia","Ethan","John","Alex","kabir","Tim"]
print("List of Customers:",customer_name)

# Store each customer's order details (customer name, product, price, category) as tuples inside a list
customer_orders = [
    ("Julia","iphone 16",90000,"Electronics"),
    ("Julia","T-Shirt", 2000, "Clothing"),
    ("Ethan","Jeans", 50, "Clothing"),
    ("John","Shoes", 800, "Footwear"),
    ("John","pen",20,"Stationary"),
    ("Alex","Suit",1000,"Clothing"),
    ("kabir","Microwave", 15000, "Home Essentials"),
    ("kabir","Vacuum Cleaner", 120, "Home Essentials"),
    ("Tim","paper", 5, "Stationary")
    ]
print("Customer Order Details:",customer_orders)

# Use a dictionary where keys are customer names and values are lists of ordered products
cust_order_dict = {}
for name in customer_name:
    cust_order_dict[name] = [order[1] for order in customer_orders if order[0] == name]

print("Customer Orders Dictionary:",cust_order_dict)
print("-" * 50)
#-----------------------------------------End of Task1---------------------------------

#Task:2 Classify products by category
  # Use a dictionary to map each product to its respective category
product_category = {order[1]: order[3] for order in customer_orders}
print("Available Product Categories:",product_category)

  # Create a set of unique product categories
unique_prod_categories = set(product_category.values())
print("Unique Product Categories:",unique_prod_categories)

  # Display all available product categories
print("Available Product Categories:", unique_prod_categories)
print("-" * 50)
#-----------------------------------------End of Task2---------------------------------

#Task:3 Analyze customer orders
  # Use a loop to calculate the total amount each customer spends
  # If the total purchase value is above $100, classify the customer as a high-value buyer
  # If it is between $50 and $100, classify the customer as a moderate buyer
  # If it is below $50, classify them as a low-value buyer

customer_spending = {}
customer_classification = {}
for name in customer_name:
    total_spent = sum(order[2] for order in customer_orders if order[0] == name)
    if total_spent > 100:
        customer_spending[name] = (total_spent, "High-Value Buyer")
        customer_classification[name] = "High-Value Buyer"
    elif 50 <= total_spent <= 100:
        customer_spending[name] = (total_spent, "Moderate Buyer")
        customer_classification[name] = "Moderate Buyer"
    else:
        customer_spending[name] = (total_spent, "Low-Value Buyer")
        customer_classification[name] = "Low-Value Buyer"
print("Customer Spending & Classification:")
for name in customer_name:
    print(f"{name}: ${customer_spending[name][0]} - {customer_classification[name]}")
print("-" * 50)
#-----------------------------------------End of Task3---------------------------------

#Task:4 Generate business insights
# Calculate the total revenue per product category and store it in a dictionary
# Extract unique products from all orders using a set
# Use a list comprehension to find all customers who purchased electronics
# Identify the top three highest-spending customers using sorting
category_revenue = {}
for category in unique_prod_categories:
    category_revenue[category] = sum(order[2] for order in customer_orders if order[3] == category)
print("Total Revenue per Product Category:")
print(category_revenue)
unique_products = set(order[1] for order in customer_orders)
print("Unique Products from All Orders:")
print(unique_products)
electronics_buyers = [order[0] for order in customer_orders if order[3] == "Electronics"]
print("Customers who purchased Electronics:")
print(electronics_buyers)
top_spenders = sorted(customer_spending.items(), key=lambda x: x[1][0], reverse=True)[:3]
print("Top Three Highest-Spending Customers:")
for name, (amount, classification) in top_spenders:
    print(f"{name}: ${amount} - {classification}")
print("-" * 50)    

 #-----------------------------------------End of Task4---------------------------------
 
#  Task:5 Organize and display data
# Print a summary of each customer’s total spending and their classification

print("Customer Spending Summary:")
for name in customer_name:
    print(f"{name}: ${customer_spending[name]} - {customer_classification[name]}")

print("-" * 50)

#  Use set operations to find customers who purchased from multiple categories
cust_categories = {}
for name in customer_name:
    cust_categories[name] = {order[3] for order in customer_orders if order[0] == name}

multi_category_customers = [name for name, cats in cust_categories.items() if len(cats) > 1]

print("Customers who purchased from multiple categories:")
print(multi_category_customers)

print("-" * 50)

# 3. Identify common customers who bought both Electronics and Clothing
electronics_buyers = {order[0] for order in customer_orders if order[3] == "Electronics"}
clothing_buyers = {order[0] for order in customer_orders if order[3] == "Clothing"}

common_customers = electronics_buyers.intersection(clothing_buyers)

print("Customers who bought both Electronics and Clothing:")
print(common_customers)
print("-" * 50)

#-----------------------------------------End of Task5---------------------------------
```
